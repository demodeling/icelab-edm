```python
%%capture
# --- Output column names: 8 #https://github.com/european-modelling-hubs/RespiCast-SyndromicIndicators/wiki/Submission-format
res_cols = [
    'origin_date',    'target',    'target_end_date',    'horizon',
    'location', 'output_type', 'output_type_id', 'value'
]


url = 'https://github.com/european-modelling-hubs/RespiCast-SyndromicIndicators/blob/main/supporting-files/forecasting_weeks.csv?raw=true'
dfv = pd.read_csv(url, index_col=0)

target_date = dfv.iloc[2,3]
current_date = dfv.index[2]

filename = f"{current_date}-IceLab-EDM.csv"

# Run ARI incidence
csv_file = 'https://github.com/european-modelling-hubs/RespiCast-SyndromicIndicators/blob/main/target-data/latest-ARI_incidence.csv?raw=true'

df = pd.read_csv(csv_file)
locs = np.unique(df['location'])

df_res_ARI = res.run_analysis(filename,df,locs, dfv,origin_date = current_date, target_incidence=df['target'].to_numpy()[0])

# Now ILI
csv_file = 'https://github.com/european-modelling-hubs/RespiCast-SyndromicIndicators/blob/main/target-data/latest-ILI_incidence.csv?raw=true'

df = pd.read_csv(csv_file)
locs = np.unique(df['location'])

df_res_ILI = res.run_analysis(filename,df,locs, dfv,origin_date = current_date, target_incidence=df['target'].to_numpy()[0])
```
